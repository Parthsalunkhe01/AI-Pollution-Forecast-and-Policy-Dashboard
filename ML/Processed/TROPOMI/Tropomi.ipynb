```python
print(df.shape)
print(df.dtypes)
df.isnull().sum()
```

```python
before = len(df)
df = df.drop_duplicates()
print("Duplicates removed:", before - len(df))
```

```python
df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
df['datetime'].isnull().sum()   # check if any invalid timestamps
```

```python
print("Missing values per column:")
df.isnull().sum()
```

```python
mean_val = df['mean_concentration'].mean()
df['mean_concentration'].fillna(mean_val, inplace=True)
print("Filled missing mean_concentration with:", mean_val)
```

```python
Q1 = df['mean_concentration'].quantile(0.25)
Q3 = df['mean_concentration'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower bound:", lower)
print("Upper bound:", upper)

outliers = df[(df['mean_concentration'] < lower) | (df['mean_concentration'] > upper)]
print("Total outliers found:", len(outliers))
```

```python
df_no_outliers = df[(df['mean_concentration'] >= lower) & (df['mean_concentration'] <= upper)]
print("Shape after removing outliers:", df_no_outliers.shape)
```

```python
df['mean_concentration'] = np.where(df['mean_concentration'] < lower, lower, df['mean_concentration'])
df['mean_concentration'] = np.where(df['mean_concentration'] > upper, upper, df['mean_concentration'])

print("Outliers capped.")
```
