```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv(r"B:\SIH2\AI-Pollution-Forecast-and-Policy-Dashboard\ML\Raw\TROPOMI\tropomi_no2_hourly_2023_2025_real.csv")
```

```python
df
```

```python
print(df.shape)
print(df.dtypes)
df.isnull().sum()
```

```python
no2 = df['mean_concentration']
```

```python
df['no2_normalized'] = (no2 - no2.min()) / (no2.max() - no2.min())
```

```python
df['no2_normalized']
```

```python
df['hour'] = pd.to_datetime(df['datetime']).dt.hour
```

```python
df['traffic_peak'] = df['hour'].apply(
    lambda h: 1 if (7 <= h <= 10) or (17 <= h <= 20) else 0
)
```

```python
df
```

```python
df['traffic_no2'] = df['no2_normalized'] * df['traffic_peak']
df['industrial_no2'] = df['no2_normalized'] * (1 - df['traffic_peak'])
```

```python
df['data_source'].value_counts()
```

```python
df.to_csv("Tropomi_2023_2025.csv",index=False)
```
